```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
c = 3e8  # Speed of light in m/s
nM, nchi = np.linspace(1, 1000, 100), np.linspace(0, 1, 100)
M, chi = np.meshgrid(nM, nchi) # Mass range in solar masses
A = (1 + np.sqrt(1 - chi ** 2)) * (M * 1.989e30 / c) ** 2 # Kerr black hole area formula

plt.contourf(M, chi, A)
plt.colorbar(label='Area (m²)')
plt.xlabel(r'$M (M_{\odot})$')
plt.ylabel(r'$\chi$')
```
